```python
#import frameworks
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
# load dataset
df=pd.read_csv(r"C:\Users\PC\Downloads\Salary_Data.csv")
```

```python
df.head(5)
```

```python
#determine features and varaibles
x=df[["YearsExperience"]]
y=df["Salary"]
```

```python
# splitting the data set
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
# train the model
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
# prediction
y_predict=model.predict(x_test)
print(y_predict)
```

```python
# evaluate model
mse=mean_squared_error(y_test,y_predict)
print(mse)
```

```python
#make prediction
experience=[[6]]
predicted_salary=model.predict(experience)
print(predicted_salary[0])
```

```python
YearsExperience=[[6]]
predicted_salary=model.predict(YearsExperience)
print(predicted_salary[0])
```

```python
# visualize the regression line
import matplotlib.pyplot as plt
plt.scatter(x,y,color="red",label="Actual Data")
plt.plot(x, model.predict(x),color="blue",linewidth=2,label="Regression line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Salary Prediction based on Years of Experience")
plt.show()
```

```python
# interpretation of the slope and intercept
print("Slope:",model.coef_[0])
print("Intercept:",model.intercept_)
```

```python
# interpretation
print("slope: For every additional years of experience,the predicted salary increases by the value of the slope coefficient.")
print("Intercept: This is the predicted salary when years of experience is 0.")
```
